```python
import numpy as np
import pandas as pd

from src.data_utils import fetch_hourly_weather_range
```

# Fetching 2000-2003

```python
# Variables to fetch
HOURLY_VARS = [
    "temperature_2m",
    "relative_humidity_2m",
    "dew_point_2m",
    "apparent_temperature",
    "precipitation",
    "pressure_msl",
    "wind_speed_10m",
    "wind_direction_10m",
    "cloud_cover",
    "shortwave_radiation",
    "wet_bulb_temperature_2m",
]
```

```python
df_00_03 = fetch_hourly_weather_range(
    start_year=2000,
    end_year=2003,
    hourly_vars=HOURLY_VARS,
    save_path=None
)
```

```python
df_00_03.tail()
```

# 00-03 Modifications

```python
adelaide_04_24_df = pd.read_csv(
    "data/processed/adelaide_hourly_weather_cleaned.csv",
    parse_dates=["datetime_utc"],
    index_col="datetime_utc"
)
adelaide_04_24_df["datetime_local"] = adelaide_04_24_df.index.tz_convert("Australia/Adelaide")
```

```python
adelaide_04_24_df.head()
```

```python
df_00_03.index.rename('datetime_utc', inplace=True)
```

```python
# Making index timezone aware
df_00_03.index = df_00_03.index.tz_localize("UTC")
df_00_03.index.tz
```

```python
# Util date columns
df_00_03["datetime_local"] = df_00_03.index.tz_convert("Australia/Adelaide")
df_00_03["date_local"] = df_00_03["datetime_local"].dt.date
df_00_03["hour_local"] = df_00_03["datetime_local"].dt.hour
df_00_03["month_local"] = df_00_03["datetime_local"].dt.month
df_00_03[['datetime_local', 'date_local', 'hour_local', 'month_local']].head()
```

```python
# Wind direction columns
from src.general_utils import wind_to_compass8

df_00_03['wind_direction_compass'] = wind_to_compass8(df_00_03['wind_direction_10m'])
df_00_03['wind_unit_x'] = np.cos(np.deg2rad(df_00_03['wind_direction_10m'])) 
df_00_03['wind_unit_y'] = np.sin(np.deg2rad(df_00_03['wind_direction_10m']))
df_00_03[['wind_direction_10m', 'wind_direction_compass', 'wind_unit_x', 'wind_unit_y']].head()
```

# Merging 00-03 with 04-24

```python
df_00_03.index[-1], adelaide_04_24_df.index[0]
```

```python
adelaide_hourly_clean = pd.concat([df_00_03, adelaide_04_24_df], axis=0)
adelaide_hourly_clean.head()
```

```python
adelaide_hourly_clean.tail()
```

# Adding Heatwave Columns to 00-24

Adding `'DMT', 'tmin', 'tmax', 'EHI_sig', 'EHI_accl', 'EHF', 'EHF_sev', 'severity'` columns from *"temperature_1958_2024.csv"* to original *2000-2024* clean data.

```python
temp_df = pd.read_csv('data/processed/temperature_1958_2024.csv',
    parse_dates=["datetime_utc"],
    index_col="datetime_utc"
)
temp_df["datetime_local"] = temp_df.index.tz_convert("Australia/Adelaide")
temp_df.head()
```

```python
adelaide_hourly_clean = adelaide_hourly_clean.join(temp_df.drop(columns=['temperature_2m', 'datetime_local']), how='left')
adelaide_hourly_clean.head()
```

```python
# Overwriting on old clean ready data since this is the current last clean version
adelaide_hourly_clean.to_csv('data/processed/adelaide_hourly_weather_cleaned.csv', index=True, index_label="datetime_utc")
```

# El Nino Southern Oscilation (ENSO) Index  Preprocesssing

```python
# Load the text file
# We assign column names since your snippet doesn't have a header
soi = pd.read_csv('data/raw/soi_monthly.txt', names=['Date', 'Value'])

# turn '187601' into an actual datetime (1876-01-01)
soi['Date'] = pd.to_datetime(soi['Date'], format='%Y%m')
soi.head()
```

```python
soi_yearly = soi.groupby(soi['Date'].dt.year).mean()
soi_yearly.head()
```

```python
soi_yearly['Date'] = soi_yearly.index
soi_yearly.rename(columns={'Date': 'year', 'Value': 'SOI'}, inplace=True)
soi_yearly.head()
```

```python
soi_yearly.to_csv('data/processed/soi_yearly.csv', index=False)
```
